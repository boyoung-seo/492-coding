> **Re:** Q4 price for our house-brand ground coffee (12 oz)
>
> The board keeps asking whether we should **cut the price** of our house-brand coffee to grow unit sales. Some managers think a lower price would pay for itself; others think we'd just give away margin.
>
> You have three years of weekly sales from all 40 stores. **Recommend one shelf price for all stores for Q4 2026** (13 weeks, Sep 28 – Dec 27).
