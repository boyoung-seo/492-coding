## Stage 1 · The business problem

> **From:** Priya Raman, VP of Retention, FreshCrate (meal-kit subscriptions)
> **Re:** Should we keep sending win-back discounts?
>
> When a customer looks like they're about to cancel, our retention agents send a **25%-off-for-8-weeks** offer. It isn't cheap, and half the exec team thinks it doesn't work. Somebody even showed a chart suggesting customers who get the discount cancel *more*.
>
> Next quarter I want to replace agent judgment with a rule. Our churn model gives every customer a `risk_score` from 1 to 100. **Recommend a threshold T: every customer with risk_score ≥ T gets the offer automatically** (T = 101 means nobody gets it).
>
> The economics:
> - Keeping a customer who would otherwise have cancelled is worth **$800** in future margin.
> - The discount costs us **$90** for a customer who stays, and **$20** for one who cancels anyway.
>
> Next quarter covers about **30,000** active customers.

### The data
