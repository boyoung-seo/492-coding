>
> **CareBridge** assigns a nurse coordinator to a patient for 30 days after discharge: follow-up calls, medication checks, appointment scheduling. It costs **$600 per patient**. Every 30-day readmission costs us about **$15,000** (care we aren't paid for, plus Medicare penalties).
>
> The CFO says the numbers show CareBridge patients are readmitted **more** often, and wants to scrap it.
>
> Next year we'll discharge about **20,000** patients. I need a rule: **enroll every patient whose `severity_score` is at least S** (S = 101 means scrap the program).
